```python
!pip install yfinance plotly

import yfinance as yf
import pandas as pd
import plotly.graph_objects as go
```

```python
nvda = yf.Ticker("NVDA")
nvda_data = nvda.history(period="1y")
```

```python
nvda_data["SMA_50"] = nvda_data["Close"].rolling(window=50).mean()
nvda_data["SMA_200"] = nvda_data["Close"].rolling(window=200).mean()
```

```python
from plotly.subplots import make_subplots
import pandas as pd

fig = make_subplots(
    rows=2, cols=1,
    shared_xaxes=True,
    vertical_spacing=0.1,
    row_heights=[0.7, 0.3],
    subplot_titles=("NVIDIA Stock: Scandal Dip vs. Long-Term Trend", "Trading Volume Around the Dip")
)

fig.add_trace(
    go.Scatter(x=nvda_data.index, y=nvda_data["Close"], name="NVDA Close", line=dict(color="blue")),
    row=1, col=1
)

fig.add_trace(
    go.Scatter(x=nvda_data.index, y=nvda_data["SMA_50"], name="50-Day SMA", line=dict(color="orange")),
    row=1, col=1
)

fig.add_trace(
    go.Scatter(x=nvda_data.index, y=nvda_data["SMA_200"], name="200-Day SMA", line=dict(color="green")),
    row=1, col=1
)

fig.add_trace(
    go.Bar(x=nvda_data.index, y=nvda_data["Volume"], name="Volume", marker_color="gray"),
    row=2, col=1
)

# Convert to epoch milliseconds to avoid the Plotly annotation sum() TypeError
target_date_2 = pd.to_datetime("2026-10-15").tz_localize(nvda_data.index.tz).timestamp() * 1000

fig.add_vline(x=target_date_2, line_dash="dash", line_color="red", annotation_text="CEO news break", row=1, col=1)
fig.add_vline(x=target_date_2, line_dash="dash", line_color="red", row=2, col=1)

fig.update_layout(
    height=700,
    showlegend=True,
    xaxis2_title="Date",
    yaxis_title="Price (USD)",
    yaxis2_title="Volume"
)

fig.show()
```
